# ABC/XYZ-анализ в денежном выражении

**Цель:** Провести ABC/XYZ-анализ по денежным показателям на всём датасете; построить сводные таблицы и тепловые карты по категориям.

**Вход:** `df_final_себестоимость_выручка_ABC_XYZ_кол.xlsx` (не приложен).

**Основные шаги:** агрегация выручки, кумулятивная доля, ABC-классификация, коэффициент вариации, XYZ-классификация, тепловые карты.

**Навыки:** pandas, numpy, seaborn, ABC/XYZ.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
valtec_df_common = pd.read_excel('df_final_себестоимость_выручка_ABC_XYZ_кол.xlsx')
```

```python
# Фильтруем данные по 2024 году
df_2024 = valtec_df_common[valtec_df_common['Год'] == 2024].copy()

# Создаем объединённую категорию
df_2024['ABC_XYZ_2024'] = df_2024['ABC_2024'] + df_2024['XYZ_2024']

# Группируем по объединённой категории
summary_2024 = df_2024.groupby('ABC_XYZ_2024').agg(
    Количество_позиций=('Код', 'nunique'),
    Сумма_остатков=('Начальный_остаток', 'sum'),
    Сумма_расхода=('Расход', 'sum'),
    Сумма_выручки=('Выручка', 'sum')
).reset_index()

# Сортируем по названию категории (по желанию)
summary_2024 = summary_2024.sort_values(by='ABC_XYZ_2024')

summary_2024
```
